```python
# Load the datasets
spx_data = pd.read_csv('archive/SPX_2007_2020.csv')
news_data = pd.read_csv('archive/year_sorted_cnbc.csv')  # Replace with your actual path

# Convert the 'published_at' column to datetime format
spx_data['date'] = pd.to_datetime(spx_data['date'])
news_data['published_at'] = pd.to_datetime(news_data['published_at'])

# Now, remove the time component and keep only the date
news_data['published_at'] = news_data['published_at'].dt.date

# Convert both columns to datetime format
spx_data['date'] = pd.to_datetime(spx_data['date'])
news_data['published_at'] = pd.to_datetime(news_data['published_at'])

# Merge datasets on common dates
common_dates = pd.merge(spx_data, news_data, left_on='date', right_on='published_at', how='inner')

# Save the combined dataset with common dates to a new CSV
common_dates.to_csv('combined_dataset_common_dates.csv', index=False)

# Display the first few rows of the 'published_at' column to confirm
print(common_dates.head())

```

```python
# Drop the 'published_at' and 'token_lengths' columns from the merged dataset
common_dates = common_dates.drop(columns=['published_at', 'token_lengths'])

# Display the first few rows of the cleaned dataset to confirm
common_dates.head()

# Save dataset to a new CSV
common_dates.to_csv('archive/combined_dataset_common_dates.csv', index=False)
```
