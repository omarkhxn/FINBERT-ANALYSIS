# Loading the Dataset

Printing existing columns to better look at what features are needed and which can be extracted and used for the analysis.

```python
# Load the dataset
df = pd.read_csv('archive/cnbc_news_datase.csv')

# Print all column names
print(df.columns)
```

## Dropping features
Dropped uncessary features and saving to a new dataset and printing the head to confirm it was done.

```python
# Drop unnecessary columns
columns_to_drop = ['index', 'url', 'short_description', 'raw_description', 'author', 'publisher', 'header_image', 'scraped_at']
df_cleaned = df.drop(columns=columns_to_drop)

# Save the cleaned dataframe as a new CSV file
df_cleaned.to_csv('archive/cleaned_cnbc_news_data.csv', index=False)
```

## Handling Missing Values

Checking for missing values in the dataset that would be modified or deleted depeding on how it affects the analysis and data for tuning

```python
missing_values = df_cleaned.isnull().sum()
print(missing_values)
```

```python
df_cleaned.dropna(subset=['description'], inplace=True)
missing_values = df_cleaned.isnull().sum()
print(missing_values)
```

## Lower case modification
changing the dataset to lowercase since the model being used is uncased and to prevent any overfitting

```python
df_cleaned['title'] = df_cleaned['title'].str.lower()
df_cleaned['description'] = df_cleaned['description'].str.lower()
df_cleaned['keywords'] = df_cleaned['keywords'].str.lower()
```

## Removing Unccessary Text
Removing texts that would affect fine tuning like URLs and symbols such as punctuations. 

```python
def clean_text(text):
    text = re.sub(r'http\S+', '', text)  # Remove URLs
    # text = re.sub(r'\d+', '', text)      # Remove numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # Remove punctuations and non-alphabetic characters
    return text

df_cleaned['title'] = df_cleaned['title'].apply(clean_text)
df_cleaned['description'] = df_cleaned['description'].apply(clean_text)
df_cleaned['published_at'] = pd.to_datetime(df['published_at'])
```
