```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the df
df = pd.read_csv('archive/combined_df_common_dates.csv')

# Distribution of Labels
plt.figure(figsize=(10, 6))
df['label'].value_counts().plot(kind='bar', color=['skyblue', 'salmon', 'lightgreen'])
plt.title('Distribution of News Sentiment Labels')
plt.xlabel('Sentiment Label')
plt.ylabel('Number of Articles')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Statistical Overview of close and volume columns
stat_overview = df[['close', 'volume']].describe()

stat_overview
```

```python
# Convert the 'date' column to datetime format for proper plotting
df['date'] = pd.to_datetime(df['date'])

# Plotting the closing prices over time
plt.figure(figsize=(14, 7))
plt.plot(df['date'], df['close'], label='Closing Prices', color='blue', alpha=0.7)

# Highlighting days with positive and negative news
plt.scatter(df[df['label'] == 'positive']['date'], df[df['label'] == 'positive']['close'], 
            color='green', label='Positive News', s=50, alpha=0.7)
plt.scatter(df[df['label'] == 'negative']['date'], df[df['label'] == 'negative']['close'], 
            color='red', label='Negative News', s=50, alpha=0.7)

plt.title('Closing Prices of SPX Over Time with News Sentiment Events')
plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.legend()
plt.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```
